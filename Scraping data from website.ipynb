```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
# 1. URL Baru yang ingin Anda scraping
url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_the_United_States_by_revenue'

# 2. Header wajib agar diizinkan oleh Robot Policy Wikipedia
headers = {
    'User-Agent': 'DataScraperBot/1.0 (ahmadhavid05@gmail.com) Python-Requests/2.0'
}

# 3. Mengambil data dari website
page = requests.get(url, headers=headers)
soup = BeautifulSoup(page.text, 'html.parser')

# 4. Mencari seluruh tabel yang ada di halaman tersebut
tabel_tabel = soup.find_all('table', class_='wikitable')

print(f"Berhasil menemukan {len(tabel_tabel)} tabel di halaman tersebut.")
```

```python
from io import StringIO
import pandas as pd

# Bungkus page.text dengan StringIO sebelum dimasukkan ke read_html
daftar_tabel = pd.read_html(StringIO(page.text))

# Mengambil tabel pertama (Daftar Perusahaan Publik Terbesar)
df_perusahaan_publik = daftar_tabel[0]

# Tampilkan 5 data teratas (Aman, rapi, dan bebas peringatan merah)
df_perusahaan_publik.head()
```
